# The Sieve of Eratosthenes

The function primes takes a list of `Numbers` that starts with some prime number `k`.
The list `Numbers` has the following properties:
* It is sorted increasingly.
* It does not contain any multiples of primes that are less than `k`. 

From the last property we know that `k`is a prime number.  

The function returns a list of all prime numbers from the list `Numbers`.

```python
def primes(Numbers: list[int]) -> list[int]:
    match Numbers:
        case []    : return []
        case p, *Ns: return [p] + primes([k for k in Ns if k % p != 0])
    return None  # type: ignore
```

```python
%%time
Ps = primes(list(range(2, 10000)))
```

```python
for p in Ps: # type: ignore
    print(p, end=', ')
```
